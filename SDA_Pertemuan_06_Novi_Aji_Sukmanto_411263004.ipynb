```python
"""
Praktik Mandiri 6.1 - Rekursi Dasar
Setiap nomor dipisah dengan komentar agar mudah dipelajari.
"""

# ==========================================================
# NOMOR 1 - Fungsi countdown(n), dijalankan untuk n = 5
# ==========================================================
def countdown(n):
    # BASE CASE: jika n sudah 0, hentikan rekursi dan tampilkan "Selesai!"
    if n == 0:
        print("Selesai!")
        return
    # Tampilkan angka saat ini
    print(n)
    # RECURSIVE CASE: panggil diri sendiri dengan n yang lebih kecil (n - 1)
    countdown(n - 1)


print("=== Nomor 1: countdown(5) ===")
countdown(5)   # Output: 5 4 3 2 1 Selesai!


```

def countdown(n) merupakan fungsi countdown dengan parameter n.



if n < 0 merupakan base case atau kondisi berhenti.



return menghentikan pemanggilan fungsi.



print(n) menampilkan angka yang sedang diproses.



countdown(n - 1) memanggil fungsi kembali dengan nilai dikurangi satu.

```python
# ==========================================================
# NOMOR 2 - Tampilkan "Masuk n" sebelum rekursi dan "Keluar n" setelahnya
# ==========================================================
def countdown_masuk_keluar(n):
    # BASE CASE: berhenti saat n == 0
    if n == 0:
        print("Selesai!")
        return
    print("Masuk", n)                 # dicetak SEBELUM pemanggilan rekursif
    countdown_masuk_keluar(n - 1)     # pemanggilan rekursif
    print("Keluar", n)                # dicetak SETELAH pemanggilan rekursif kembali
    # Catatan: "Keluar" tercetak dari angka kecil ke besar (1,2,3,...)
    # karena fungsi yang dipanggil paling akhir selesai paling dulu (stack LIFO).


print("\n=== Nomor 2: countdown dengan Masuk/Keluar ===")
countdown_masuk_keluar(5)


# ==========================================================
# NOMOR 3 - recursive_sum_linear(data, index=0) tanpa slicing
# ==========================================================
def recursive_sum_linear(data, index=0):
    # BASE CASE: index sudah melewati elemen terakhir -> tidak ada lagi yang dijumlahkan
    if index == len(data):
        return 0
    # RECURSIVE CASE: elemen saat ini + jumlah sisa elemen (mulai index + 1)
    # Tidak memakai slicing (data[1:]), hanya menggeser index.
    return data[index] + recursive_sum_linear(data, index + 1)
```

```python
# ==========================================================
# NOMOR 4 - Uji list kosong, satu elemen, dan lima elemen
# ==========================================================
print("\n=== Nomor 4: Pengujian recursive_sum_linear ===")
print("List kosong   [] ->", recursive_sum_linear([]))                  # 0
print("Satu elemen   [7] ->", recursive_sum_linear([7]))                # 7
print("Lima elemen   [1,2,3,4,5] ->", recursive_sum_linear([1, 2, 3, 4, 5]))  # 15
```

```python
# ==========================================================
# NOMOR 5 - Dry run untuk input [4, 6, 2]
# (versi yang mencetak jejak eksekusi agar sesuai tabel dry run)
# ==========================================================
def recursive_sum_trace(data, index=0, depth=0):
    spasi = "  " * depth   # indentasi untuk menunjukkan kedalaman rekursi
    print(f"{spasi}panggil recursive_sum_linear(data, index={index})")
    if index == len(data):
        print(f"{spasi}index == len(data) -> BASE CASE, return 0")
        return 0
    sisa = recursive_sum_trace(data, index + 1, depth + 1)
    hasil = data[index] + sisa
    print(f"{spasi}return {data[index]} + {sisa} = {hasil}")
    return hasil


print("\n=== Nomor 5: Dry run [4, 6, 2] ===")
total = recursive_sum_trace([4, 6, 2])
print("Hasil akhir:", total)   # 12

"""
DRY RUN (tertulis) untuk data = [4, 6, 2], len(data) = 3

Fase turun (pemanggilan):
  f(index=0) -> 4 + f(1)
  f(index=1) -> 6 + f(2)
  f(index=2) -> 2 + f(3)
  f(index=3) -> index == len(data) -> return 0   (BASE CASE)

Fase naik (nilai kembali):
  f(3) = 0
  f(2) = 2 + 0 = 2
  f(1) = 6 + 2 = 8
  f(0) = 4 + 8 = 12

Hasil akhir = 12
"""
```

```python
# ==========================================================
# NOMOR 6 - Base case dan recursive case (tertulis)
# ==========================================================
"""
Untuk countdown(n):
  Base case      : n == 0  -> cetak "Selesai!" lalu berhenti (return).
  Recursive case : n > 0   -> cetak n, lalu panggil countdown(n - 1).

Untuk recursive_sum_linear(data, index):
  Base case      : index == len(data)  -> return 0
                   (sudah tidak ada elemen tersisa; list kosong juga langsung
                    masuk ke sini sehingga hasilnya 0).
  Recursive case : index < len(data)   -> return data[index] + recursive_sum_linear(data, index + 1)
                   (jumlah = elemen saat ini + jumlah elemen sisanya).

Setiap pemanggilan rekursif selalu mendekati base case (n berkurang / index bertambah),
sehingga rekursi pasti berhenti.
"""
```

```python
# ==========================================================
# NOMOR 7 - Time complexity dan space complexity
# ==========================================================
"""
Fungsi recursive_sum_linear (n = jumlah elemen list):

  Time complexity  : O(n)
    Fungsi dipanggil tepat n + 1 kali (index 0 sampai n), dan tiap pemanggilan
    hanya melakukan pekerjaan konstan (satu penjumlahan dan satu perbandingan).
    Karena tidak memakai slicing, tidak ada biaya penyalinan list.

  Space complexity : O(n)
    Tidak membuat struktur data baru, tetapi kedalaman call stack mencapai
    n + 1 frame sebelum base case tercapai. Tiap frame menyimpan data, index,
    dan alamat kembali.

Fungsi countdown(n): time O(n), space O(n) (kedalaman stack n + 1) dengan alasan yang sama.
"""
```

```python
"""
6.26. Praktik Mandiri 6.2
FIBONACCI DAN MEMOIZATION
"""

import timeit

# =====================================================
# NOMOR 1 - STEP 1
# IMPLEMENTASI FIBONACCI NAIF
# =====================================================

# Variabel untuk menghitung jumlah pemanggilan fungsi
jumlah_naif = 0


def fibonacci_naif(n):
    global jumlah_naif

    # Setiap fungsi dipanggil, penghitung bertambah 1
    jumlah_naif += 1

    # STEP 1: Kondisi dasar (base case)
    if n <= 1:
        return n

    # STEP 2: Pemanggilan fungsi secara rekursif
    return fibonacci_naif(n - 1) + fibonacci_naif(n - 2)


# =====================================================
# NOMOR 1 - STEP 2
# IMPLEMENTASI FIBONACCI DENGAN MEMOIZATION
# =====================================================

# Variabel untuk menghitung pemanggilan fungsi
jumlah_memo = 0

# Dictionary untuk menyimpan hasil Fibonacci
memo = {}


def fibonacci_memo(n):
    global jumlah_memo

    # Menghitung setiap pemanggilan fungsi
    jumlah_memo += 1

    # STEP 1: Jika hasil sudah tersedia di memo,
    # langsung kembalikan hasilnya
    if n in memo:
        return memo[n]

    # STEP 2: Kondisi dasar
    if n <= 1:
        return n

    # STEP 3: Menghitung Fibonacci secara rekursif
    # dan menyimpan hasilnya ke dictionary
    hasil = fibonacci_memo(n - 1) + fibonacci_memo(n - 2)

    memo[n] = hasil

    return hasil

# =====================================================
# NOMOR 2 - STEP 3
# MENGUJI KEDUA FUNGSI
# UNTUK n = 5, 10, 20, DAN 30
# =====================================================

print("=" * 65)
print("NOMOR 2: PENGUJIAN FIBONACCI NAIF DAN MEMOIZATION")
print("=" * 65)

daftar_n = [5, 10, 20, 30]

for n in daftar_n:

    # Reset penghitung Fibonacci Naif
    jumlah_naif = 0

    # Menghitung Fibonacci Naif
    hasil_naif = fibonacci_naif(n)
    pemanggilan_naif = jumlah_naif

    # Reset penghitung dan memo
    jumlah_memo = 0
    memo = {}

    # Menghitung Fibonacci Memoization
    hasil_memo = fibonacci_memo(n)
    pemanggilan_memo = jumlah_memo

    # Menampilkan hasil pengujian
    print(f"\nNilai n = {n}")
    print(f"Hasil Fibonacci Naif       : {hasil_naif}")
    print(f"Hasil Fibonacci Memoization: {hasil_memo}")
    print(f"Pemanggilan fungsi Naif    : {pemanggilan_naif}")
    print(f"Pemanggilan fungsi Memo    : {pemanggilan_memo}")
```

```python
# =====================================================
# NOMOR 3 - STEP 4
# MENAMBAHKAN PENGHITUNG PEMANGGILAN FUNGSI
# =====================================================

print("\n" + "=" * 65)
print("NOMOR 3: PENGHITUNG JUMLAH PEMANGGILAN FUNGSI")
print("=" * 65)

n = 20

jumlah_naif = 0
hasil_naif = fibonacci_naif(n)

jumlah_memo = 0
memo = {}
hasil_memo = fibonacci_memo(n)

print(f"Pengujian untuk n = {n}")
print(f"Jumlah pemanggilan Naif : {jumlah_naif}")
print(f"Jumlah pemanggilan Memo : {jumlah_memo}")
```

```python
# =====================================================
# NOMOR 4 - STEP 5
# MEMBANDINGKAN JUMLAH PEMANGGILAN
# =====================================================

print("\n" + "=" * 65)
print("NOMOR 4: PERBANDINGAN JUMLAH PEMANGGILAN")
print("=" * 65)

print(f"{'n':<10}{'Naif':<20}{'Memoization':<20}")
print("-" * 50)

for n in daftar_n:

    jumlah_naif = 0
    fibonacci_naif(n)
    panggilan_naif = jumlah_naif

    jumlah_memo = 0
    memo = {}
    fibonacci_memo(n)
    panggilan_memo = jumlah_memo

    print(f"{n:<10}{panggilan_naif:<20}{panggilan_memo:<20}")
```

```python
# =====================================================
# NOMOR 5 - STEP 6
# MENGUKUR WAKTU EKSEKUSI DENGAN TIMEIT
# =====================================================

print("\n" + "=" * 65)
print("NOMOR 5: PENGUKURAN WAKTU EKSEKUSI")
print("=" * 65)

print(f"{'n':<10}{'Waktu Naif (detik)':<25}"
      f"{'Waktu Memo (detik)':<25}")
print("-" * 60)

for n in daftar_n:

    # Mengukur waktu Fibonacci Naif
    waktu_naif = timeit.timeit(
        lambda: fibonacci_naif(n),
        number=10
    )

    # Mengukur waktu Fibonacci Memoization
    # Menggunakan memo lokal yang baru untuk setiap pengulangan
    def jalankan_memo(n=n):
        cache = {0: 0, 1: 1}

        def hitung(k):
            if k in cache:
                return cache[k]

            cache[k] = hitung(k - 1) + hitung(k - 2)
            return cache[k]

        return hitung(n)

    waktu_memo = timeit.timeit(
        jalankan_memo,
        number=10
    )

    print(f"{n:<10}{waktu_naif:<25.8f}{waktu_memo:<25.8f}")
```

```python
# =====================================================
# NOMOR 6 - STEP 7
# PENJELASAN MANFAAT MEMOIZATION
# =====================================================

print("\n" + "=" * 65)
print("NOMOR 6: MANFAAT MEMOIZATION")
print("=" * 65)

print("""
Memoization membuat program lebih cepat karena:

1. Hasil perhitungan yang sudah pernah dilakukan
   disimpan dalam dictionary.

2. Ketika hasil yang sama dibutuhkan kembali,
   program tidak perlu menghitung ulang.

3. Fibonacci Naif melakukan banyak perhitungan
   berulang sehingga jumlah pemanggilan fungsi
   meningkat secara eksponensial.

4. Fibonacci Memoization mengurangi perhitungan
   berulang sehingga setiap nilai Fibonacci
   cukup dihitung satu kali.

Kesimpulan:
Memoization menghemat waktu pemrosesan dengan
menyimpan hasil perhitungan sebelumnya.
""")
```
