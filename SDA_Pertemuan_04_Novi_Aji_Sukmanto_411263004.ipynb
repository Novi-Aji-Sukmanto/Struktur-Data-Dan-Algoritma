```python
class Stack:
  def __init__(self):
    self.items = []
  def is_empty(self):
    return len(self.items) == 0
  def push(self, data):
    self.items.append(data)
  def pop(self):
    if self.is_empty():
      return None
    return self.items.pop()
  def peek(self):
    if self.is_empty():
      return None
    return self.items[-1]
  def size(self):
    return len(self.items)
```

```python
stack = Stack()
print(stack.is_empty())

stack.push(10)
stack.push(20)
stack.push(30)
print('Top:', stack.peek())
print('Ukuran:', stack.size())
print('Keluar:', stack.pop())
print('Top baru:', stack.peek())
```

```python
class BoundedStack:
  def __init__(self, capacity):
    self.items = []
    self.capacity = capacity

  def push(self, data):
    if len(self.items) >= self.capacity:
      return False
    self.items.append(data)
    return True

s = BoundedStack(2)
print(s.push('A'))
print(s.push('B'))
print(s.push('C'))
print(s.push('D'))
print(s.items)
```

```python
def reverse_text(text):
  stack = Stack()
  for char in text:
    stack.push(char)
  result = ''
  while not stack.is_empty():
    result += stack.pop()
  return result
print(reverse_text('DATA'))
```

```python
def is_balanced(expression):
  stack = Stack()
  pairs = {')': '(', ']': '[', '}': '{'}
  for char in expression:
    if char in '([{':
      stack.push(char)
    elif char in ')]}':
      if stack.is_empty():
        return False
      if stack.pop() != pairs[char]:
        return False
  return stack.is_empty()

print(is_balanced('(a + b) * [c - d]'))
print(is_balanced('([)]'))
print(is_balanced('((a+b)'))
```

```python
def evaluate_postfix(tokens):
  stack = Stack()
  for token in tokens:
    if token.isdigit():
      stack.push(int(token))
    else:
      b = stack.pop()
      a = stack.pop()
      if token == '+':
        stack.push(a + b)
      elif token == '-':
        stack.push(a - b)
      elif token == '*':
        stack.push(a * b)
      elif token == '/':
        stack.push(a / b)
  return stack.pop()
expr = ['2', '3', '+', '6', '-']
print(evaluate_postfix(expr))
```

```python
# ============================================
# PRAKTIK MANDIRI 4.1 - OPERASI DASAR STACK
# ============================================

# 1. Membuat class Stack menggunakan Python list
class Stack:
    def __init__(self):
        self.items = []

    # Menambahkan data ke Stack
    def push(self, data):
        self.items.append(data)

    # Mengeluarkan data paling atas
    def pop(self):
        if self.is_empty():
            return None
        return self.items.pop()

    # Melihat data paling atas
    def top(self):
        if self.is_empty():
            return None
        return self.items[-1]

    # Menghitung jumlah data
    def size(self):
        return len(self.items)

    # Mengecek apakah Stack kosong
    def is_empty(self):
        return len(self.items) == 0

    # Menampilkan isi Stack
    def display(self):
        return self.items


# ============================================
# 2. Membuat Stack dan menambahkan data
# ============================================

stack = Stack()

stack.push(10)
stack.push(20)
stack.push(30)
stack.push(40)

print("Isi Stack setelah push 10, 20, 30, 40:")
print(stack.display())


# ============================================
# 3. Menampilkan TOP dan SIZE
# ============================================

print("\nTOP  :", stack.top())
print("SIZE :", stack.size())


# ============================================
# 4. Melakukan POP dua kali
# ============================================

nilai1 = stack.pop()
nilai2 = stack.pop()

print("\nNilai yang keluar saat pop:")
print("Pop pertama :", nilai1)
print("Pop kedua   :", nilai2)

print("Isi Stack setelah dua kali pop:")
print(stack.display())


# ============================================
# 5. Menambahkan data 50 dan 60
# ============================================

stack.push(50)
stack.push(60)

print("\nSetelah menambahkan 50 dan 60:")
print(stack.display())


# ============================================
# 6. Mengeluarkan semua elemen sampai kosong
# ============================================

print("\nMengeluarkan semua elemen:")

while not stack.is_empty():
    print("Pop:", stack.pop())

print("Isi Stack setelah semua elemen dikeluarkan:")
print(stack.display())


# ============================================
# 7. Menguji POP ketika Stack kosong
# ============================================

hasil_pop = stack.pop()

print("\nUji pop ketika Stack kosong:")
print("Hasil pop:", hasil_pop)

if hasil_pop is None:
    print("Stack kosong, sehingga pop tidak mengeluarkan nilai.")
```

```python
def is_balanced(expression):
    stack = []

    # Pasangan bracket pembuka dan penutup
    pasangan = {
        ')': '(',
        ']': '[',
        '}': '{'
    }

    # Proses setiap karakter
    for char in expression:

        # Jika karakter adalah bracket pembuka
        if char in "([{":
            stack.append(char)

        # Jika karakter adalah bracket penutup
        elif char in ")]}":

            # Jika stack kosong, berarti tidak seimbang
            if not stack:
                return False

            # Ambil bracket paling atas
            top = stack.pop()

            # Periksa apakah pasangannya benar
            if top != pasangan[char]:
                return False

    # Jika stack kosong berarti semua bracket sudah berpasangan
    return len(stack) == 0
```

```python
def is_balanced(expression):
    stack = []

    # Pasangan bracket penutup dengan pembukanya
    pasangan = {
        ')': '(',
        ']': '[',
        '}': '{'
    }

    for char in expression:

        # Abaikan karakter selain bracket
        if char not in "()[]{}":
            continue

        # Jika bracket pembuka, masukkan ke stack
        if char in "([{":
            stack.append(char)

        # Jika bracket penutup
        else:
            # Stack kosong berarti tidak ada pasangan
            if not stack:
                return False

            # Ambil bracket paling atas
            top = stack.pop()

            # Periksa kecocokan pasangan
            if top != pasangan[char]:
                return False

    # Seimbang jika stack kosong
    return len(stack) == 0


# Pengujian
expressions = [
    "(a+b)",
    "([{}])",
    "([)]",
    "((a+b)",
    "{a+[b*c]}"
]

print("HASIL PENGUJIAN BALANCED BRACKETS")
print("-" * 40)

for expression in expressions:
    hasil = is_balanced(expression)

    if hasil:
        status = "Seimbang"
    else:
        status = "Tidak Seimbang"

    print(f"{expression:15} -> {status}")


# Pengujian karakter selain bracket
print("\nPENGUJIAN KARAKTER LAIN")
print("-" * 40)

print("a + (b * c)       ->", is_balanced("a + (b * c)"))
print("x + [y - {z}]     ->", is_balanced("x + [y - {z}]"))
print("a + ([b + c)]     ->", is_balanced("a + ([b + c)]"))

print("\nKompleksitas waktu  : O(n)")
print("Kompleksitas ruang  : O(n)")

# Membandingkan jumlah bracket pembuka dan penutup tidak cukup karena
# yang harus diperiksa bukan hanya jumlahnya,
# tetapi juga urutan dan pasangan bracket.
```

```python
# ==========================================
# PRAKTIK MANDIRI 4.3
# STACK BERBASIS LINKED LIST
# ==========================================

# 1. Membuat class Node
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None


# 2. Membuat class LinkedStack
class LinkedStack:
    def __init__(self):
        self.top = None
        self._size = 0

    # Operasi PUSH
    def push(self, data):
        new_node = Node(data)

        # Node baru menjadi top
        new_node.next = self.top
        self.top = new_node

        self._size += 1

    # Operasi POP
    def pop(self):
        if self.top is None:
            print("Stack kosong!")
            return None

        data = self.top.data

        # Pindahkan top ke node berikutnya
        self.top = self.top.next

        self._size -= 1

        return data

    # Operasi PEEK
    def peek(self):
        if self.top is None:
            return None

        return self.top.data

    # Mengembalikan ukuran stack
    def size(self):
        return self._size

    # Menampilkan struktur stack
    def display(self):
        current = self.top

        print("TOP", end=" -> ")

        while current is not None:
            print(current.data, end=" -> ")
            current = current.next

        print("None")


# ==========================================
# PROGRAM UTAMA
# ==========================================

stack = LinkedStack()

print("==========================================")
print("STACK BERBASIS LINKED LIST")
print("==========================================")

# 3. PUSH A
print("\n1. PUSH A")
stack.push("A")
stack.display()

# 3. PUSH B
print("\n2. PUSH B")
stack.push("B")
stack.display()

# 3. PUSH C
print("\n3. PUSH C")
stack.push("C")
stack.display()


# ==========================================
# 4. POP SEKALI
# ==========================================

print("\n==========================================")
print("OPERASI POP")
print("==========================================")

nilai_keluar = stack.pop()

print("Data yang keluar:", nilai_keluar)
print("Struktur setelah POP:")
stack.display()


# ==========================================
# 5. PEEK DAN SIZE
# ==========================================

print("\n==========================================")
print("PEEK DAN SIZE")
print("==========================================")

print("Nilai TOP (peek):", stack.peek())
print("Ukuran Stack (size):", stack.size())


# ==========================================
# INFORMASI NODE
# ==========================================

print("\n==========================================")
print("INFORMASI NODE")
print("==========================================")

current = stack.top

while current is not None:
    if current.next is not None:
        print(
            "Node", current.data,
            "-> next =", current.next.data
        )
    else:
        print(
            "Node", current.data,
            "-> next = None"
        )

    current = current.next

# Mengapa push() dan pop() adalah O(1)?
# Karena keduanya hanya melakukan beberapa perubahan pointer,
# tanpa harus mengunjungi seluruh node
```
