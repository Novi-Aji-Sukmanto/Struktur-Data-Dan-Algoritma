```python
# Praktik Mandiri 5.1 — Queue Dasar

from collections import deque


class Queue:
    """Queue (antrian) FIFO berbasis collections.deque."""

    def __init__(self):
        self._data = deque()

    def enqueue(self, item):
        """Tambah elemen di belakang (rear)."""
        self._data.append(item)

    def dequeue(self):
        """Keluarkan elemen dari depan (front)."""
        if self.is_empty():
            raise IndexError("Queue kosong: tidak bisa dequeue")
        return self._data.popleft()

    def peek(self):
        """Lihat elemen depan tanpa mengeluarkannya."""
        if self.is_empty():
            raise IndexError("Queue kosong: tidak ada front untuk di-peek")
        return self._data[0]

    def is_empty(self):
        return len(self._data) == 0

    def size(self):
        return len(self._data)

    def __str__(self):
        return "front -> " + str(list(self._data)) + " <- rear"


# ============== PROGRAM UTAMA ==============
q = Queue()
urutan_masuk = []
urutan_keluar = []

# 2. Masukkan data 10, 20, 30, 40, 50
print("=== 2. Enqueue 10, 20, 30, 40, 50 ===")
for nilai in [10, 20, 30, 40, 50]:
    q.enqueue(nilai)
    urutan_masuk.append(nilai)
    print(f"enqueue({nilai}) -> {q}")

# 3. Tampilkan front dan size
print("\n=== 3. Front dan Size ===")
print("Front :", q.peek())
print("Size  :", q.size())

# 4. Dequeue dua kali
print("\n=== 4. Dequeue dua kali ===")
for _ in range(2):
    keluar = q.dequeue()
    urutan_keluar.append(keluar)
    print(f"dequeue() -> {keluar} | sisa: {q}")

# 5. Tambahkan 60 dan 70
print("\n=== 5. Enqueue 60 dan 70 ===")
for nilai in [60, 70]:
    q.enqueue(nilai)
    urutan_masuk.append(nilai)
    print(f"enqueue({nilai}) -> {q}")

# 6. Keluarkan seluruh elemen sampai kosong
print("\n=== 6. Dequeue sampai kosong ===")
while not q.is_empty():
    keluar = q.dequeue()
    urutan_keluar.append(keluar)
    print(f"dequeue() -> {keluar} | sisa: {q}")

# 7. Uji dequeue dan peek saat kosong
print("\n=== 7. Uji Queue kosong ===")
try:
    q.dequeue()
except IndexError as e:
    print("dequeue() ->", e)

try:
    q.peek()
except IndexError as e:
    print("peek()    ->", e)

# 9. Bukti FIFO
print("\n=== 9. Bukti FIFO ===")
print("Urutan masuk  :", urutan_masuk)
print("Urutan keluar :", urutan_keluar)
print("FIFO terbukti :", urutan_masuk == urutan_keluar)
```

```python
# Praktik Mandiri 5.2 — Circular Queue

class CircularQueue:
    def __init__(self, capacity):
        self.capacity = capacity
        self.items = [None] * capacity   # array fisik
        self.front = 0                   # indeks elemen terdepan
        self.rear = -1                   # indeks elemen terakhir
        self.count = 0                   # jumlah elemen

    def is_full(self):
        return self.count == self.capacity

    def is_empty(self):
        return self.count == 0

    def enqueue(self, value):
        if self.is_full():
            return False                 # Queue penuh (overflow)
        self.rear = (self.rear + 1) % self.capacity   # wrap-around
        self.items[self.rear] = value
        self.count += 1
        return True

    def dequeue(self):
        if self.is_empty():
            return None                  # Queue kosong (underflow)
        value = self.items[self.front]
        self.items[self.front] = None
        self.front = (self.front + 1) % self.capacity  # wrap-around
        self.count -= 1
        return value

    def logical_order(self):
        """Urutan logis dari front ke rear."""
        return [self.items[(self.front + i) % self.capacity]
                for i in range(self.count)]

    def show(self, label):
        print(f"\n>> {label}")
        print(f"   items  : {self.items}")
        print(f"   front  : {self.front}")
        print(f"   rear   : {self.rear}")
        print(f"   count  : {self.count}")
        print(f"   logis  : {self.logical_order()}  (front -> rear)")


# ---------------------------------------------------------------
# Praktik Mandiri 5.2
# ---------------------------------------------------------------
# 1. Buat CircularQueue kapasitas 4
q = CircularQueue(4)
q.show("1. Queue dibuat (kapasitas 4)")

# 2. Enqueue A, B, C, D dan cek kondisi penuh
for x in ["A", "B", "C", "D"]:
    q.enqueue(x)
    q.show(f"2. Enqueue {x}")
print(f"\n   is_full() = {q.is_full()}")

# 3. Enqueue E saat penuh
hasil = q.enqueue("E")
q.show("3. Enqueue E saat penuh")
print(f"   Hasil enqueue E = {hasil} -> DITOLAK (Queue penuh/overflow), "
      "isi Queue tidak berubah")

# 4. Dequeue dua elemen
for _ in range(2):
    v = q.dequeue()
    q.show(f"4. Dequeue -> {v}")

# 5. Enqueue E dan F (rear wrap-around)
for x in ["E", "F"]:
    q.enqueue(x)
    q.show(f"5. Enqueue {x}")

# 7. Keadaan akhir array fisik dan Queue logis
print("\n" + "=" * 55)
print("7. KEADAAN AKHIR")
print("=" * 55)
print("Indeks array fisik :   0    1    2    3")
print("Isi array fisik    : ", q.items)
print("Penanda            :", "      ", "rear", "      ", "front")
print("Queue logis        : ", q.logical_order(), "(front -> rear)")

# 8 & 9. Penjelasan
print("""
8. FUNGSI MODULO (%)
   rear  = (rear + 1)  % capacity
   front = (front + 1) % capacity
   Modulo membuat indeks kembali ke 0 setelah mencapai capacity-1,
   sehingga array dipakai melingkar (wrap-around) dan slot kosong di
   depan dapat dipakai ulang tanpa menggeser elemen.
   Contoh: rear = 3 -> (3 + 1) % 4 = 0.

9. KOMPLEKSITAS
   enqueue : O(1)  (tanpa loop/pergeseran elemen)
   dequeue : O(1)
""")
```

```python
# Praktik Mandiri 5.3 - Deque dan Palindrome

from collections import deque
import re


def normalisasi(teks):
    """Huruf kecil, buang spasi dan tanda baca (sisakan huruf & angka)."""
    return re.sub(r"[^a-z0-9]", "", teks.lower())


# ---------------------------------------------------------------
# 1. Fungsi is_palindrome() menggunakan Deque
# ---------------------------------------------------------------
def is_palindrome(teks):
    d = deque(normalisasi(teks))
    while len(d) > 1:
        if d.popleft() != d.pop():   # bandingkan ujung kiri & kanan
            return False
    return True


# ---------------------------------------------------------------
# 5. Versi yang menampilkan pasangan kiri-kanan tiap iterasi
# ---------------------------------------------------------------
def is_palindrome_trace(teks):
    d = deque(normalisasi(teks))
    print(f"Hasil normalisasi : '{''.join(d)}'")
    iterasi = 0
    while len(d) > 1:
        iterasi += 1
        kiri = d.popleft()
        kanan = d.pop()
        cocok = kiri == kanan
        print(f"  Iterasi {iterasi}: kiri='{kiri}' | kanan='{kanan}' -> "
              f"{'cocok' if cocok else 'TIDAK cocok'}")
        if not cocok:
            return False
    if len(d) == 1:
        print(f"  Sisa 1 karakter di tengah: '{d[0]}' (tidak perlu dibandingkan)")
    return True


if __name__ == "__main__":
    # 2. Uji kata
    print("=== 2. Uji kata ===")
    for kata in ["level", "katak", "radar", "python", "algoritma"]:
        print(f"{kata:<10} -> {is_palindrome(kata)}")

    # 3. Uji kalimat (spasi & tanda baca)
    print("\n=== 3. Uji kalimat ===")
    kalimat = [
        "Ira Bini Bani Ari",
        "A man, a plan, a canal: Panama!",
        "Kasur ini rusak",
        "Saya suka Python",
    ]
    for k in kalimat:
        print(f"'{k}' -> normalisasi '{normalisasi(k)}' -> {is_palindrome(k)}")

    # 4 & 5. Dry run + tampilan pasangan kiri-kanan
    print("\n=== 4 & 5. Dry run 'katak' ===")
    print(is_palindrome_trace("katak"))

    print("\n=== Dry run 'python' (bukan palindrome) ===")
    print(is_palindrome_trace("python"))

    print("\n=== Dry run kalimat 'Ira Bini Bani Ari' ===")
    print(is_palindrome_trace("Ira Bini Bani Ari"))
```

6. Mengapa Deque lebih natural daripada Queue satu arah?

Memeriksa palindrome berarti membandingkan karakter dari kedua ujung secara bergantian: depan dengan belakang, lalu bergeser ke dalam. Deque (double-ended queue) mendukung popleft() dan pop() dengan cost O(1). Itu cocok dengan pola ini.

Queue satu arah hanya bisa mengambil dari depan (dequeue) dan memasukkan dari belakang (enqueue). Karakter paling belakang tidak bisa diambil langsung. Anda harus mengeluarkan semua elemen di depannya dulu, atau menyalinnya ke struktur lain seperti stack. Itu membuat kodenya lebih rumit dan boros memori.

7. Kompleksitas

Waktu: O(n). Normalisasi membaca n karakter. Loop berjalan paling banyak n/2 kali, dan tiap popleft()/pop() bernilai O(1).
Ruang: O(n). Deque menyimpan seluruh karakter hasil normalisasi, termasuk string hasil normalisasinya.
