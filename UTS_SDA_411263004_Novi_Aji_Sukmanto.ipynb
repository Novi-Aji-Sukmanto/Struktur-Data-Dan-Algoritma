## Nomor 1 - Fungsi `linear_search_nim`

```python
# ============================================================
# NOMOR 1: LINEAR SEARCH (manual)
# ============================================================
def linear_search_nim(data, target):
    """Mencari kode mahasiswa (kolom 0) dari indeks 0 sampai akhir.
    Mengembalikan (indeks, jumlah_perbandingan). Indeks = -1 jika tidak ada."""
    perbandingan = 0                      # penghitung jumlah perbandingan
    for i in range(len(data)):            # telusuri satu per satu dari awal
        perbandingan += 1                 # setiap pengecekan = 1 perbandingan
        if data[i][0] == target:          # bandingkan kode mahasiswa dengan target
            return i, perbandingan        # ketemu -> kembalikan indeks & jumlah perbandingan
    return -1, perbandingan               # sampai habis tidak ketemu
```

## Nomor 3 - `merge` dan `merge_sort_nim` (rekursif)

```python
# ============================================================
# NOMOR 3: MERGE SORT REKURSIF (manual)
# ============================================================
def merge(left, right):
    """Menggabungkan dua list yang SUDAH terurut menjadi satu list terurut
    (ascending berdasarkan kode mahasiswa di indeks 0)."""
    hasil = []                            # list penampung hasil (ini ruang tambahan O(n))
    i = 0                                 # pointer untuk list kiri
    j = 0                                 # pointer untuk list kanan

    # Selama kedua list masih punya elemen, ambil yang lebih kecil
    while i < len(left) and j < len(right):
        if left[i][0] <= right[j][0]:     # <= menjaga kestabilan (stable sort)
            hasil.append(left[i])
            i += 1
        else:
            hasil.append(right[j])
            j += 1

    # Salah satu list sudah habis; sisa list lainnya sudah terurut -> tempel semua
    while i < len(left):
        hasil.append(left[i])
        i += 1
    while j < len(right):
        hasil.append(right[j])
        j += 1
    return hasil


def merge_sort_nim(data):
    """Merge Sort rekursif. Mengembalikan list BARU (data asli tidak diubah)."""
    # BASE CASE: list kosong / 1 elemen sudah terurut
    if len(data) <= 1:
        return data[:]                    # kembalikan salinan

    # DIVIDE: bagi dua di titik tengah
    mid = len(data) // 2
    kiri = merge_sort_nim(data[:mid])     # CONQUER: urutkan separuh kiri (rekursi)
    kanan = merge_sort_nim(data[mid:])    # CONQUER: urutkan separuh kanan (rekursi)

    # COMBINE: gabungkan dua hasil terurut
    return merge(kiri, kanan)

print("Fungsi merge dan merge_sort_nim siap dipakai.")
```

## Nomor 5 - Fungsi `binary_search_nim`

```python
# ============================================================
# NOMOR 5: BINARY SEARCH (manual, iteratif)
# ============================================================
def binary_search_nim(data_urut, target):
    """Binary Search pada data yang SUDAH terurut berdasarkan kode.
    Mengembalikan (indeks, jumlah_perbandingan). Indeks = -1 jika tidak ada.
    Catatan: setiap iterasi (satu kali membandingkan target dengan elemen tengah)
    dihitung sebagai 1 perbandingan."""
    low = 0                               # batas bawah pencarian
    high = len(data_urut) - 1             # batas atas pencarian
    perbandingan = 0

    while low <= high:                    # selama ruang pencarian masih ada
        mid = (low + high) // 2           # titik tengah
        perbandingan += 1                 # 1 perbandingan dengan elemen tengah
        kode_tengah = data_urut[mid][0]

        if kode_tengah == target:         # ketemu
            return mid, perbandingan
        elif kode_tengah < target:        # target lebih besar -> buang separuh kiri
            low = mid + 1
        else:                             # target lebih kecil -> buang separuh kanan
            high = mid - 1

    return -1, perbandingan               # ruang pencarian habis -> tidak ada
```

## Nomor 7 - Trace / Dry Run Binary Search (opsi b)

```python
# ============================================================
# NOMOR 7: TRACE BINARY SEARCH untuk target_nim
# ============================================================
def binary_search_trace(data_urut, target):
    """Sama seperti binary_search_nim, tetapi mencetak perubahan low/high/mid."""
    low, high = 0, len(data_urut) - 1
    langkah = 0
    print(f"Target: {target}")
    print(f"{'Langkah':<8}{'low':<5}{'high':<6}{'mid':<5}{'kode[mid]':<11}Keputusan")
    while low <= high:
        langkah += 1
        mid = (low + high) // 2
        kode = data_urut[mid][0]
        if kode == target:
            keputusan = "SAMA -> ditemukan"
        elif kode < target:
            keputusan = f"{kode} < {target} -> low = {mid + 1}"
        else:
            keputusan = f"{kode} > {target} -> high = {mid - 1}"
        print(f"{langkah:<8}{low:<5}{high:<6}{mid:<5}{kode:<11}{keputusan}")
        if kode == target:
            return mid
        if kode < target:
            low = mid + 1
        else:
            high = mid - 1
    print("Ruang pencarian habis -> tidak ditemukan")
    return -1

binary_search_trace(data_urut, target_nim)
```

## Nomor 9 - Analisis (Text cell)

**Kompleksitas**
| Algoritma | Best | Average | Worst |
|---|---|---|---|
| Linear Search | O(1) | O(n) | O(n) |
| Binary Search | O(1) | O(log n) | O(log n) |
| Merge Sort | O(n log n) | O(n log n) | O(n log n) |

- **Linear Search**: memeriksa elemen satu per satu; pada kasus terburuk (data di akhir / tidak ada) dilakukan n perbandingan. Untuk n = 12, kasus tidak ada = 12 perbandingan.
- **Binary Search**: setiap langkah membuang separuh ruang pencarian, sehingga maksimal sekitar log2(n)+1 perbandingan (untuk n = 12 maksimal 4).
- **Merge Sort**: data dibagi dua secara rekursif sebanyak log n tingkat, dan tiap tingkat melakukan penggabungan total n elemen, sehingga O(n log n).

**Ruang tambahan Merge Sort**: O(n), karena fungsi `merge` membuat list `hasil` baru dan pemotongan list menghasilkan salinan; ditambah kedalaman rekursi O(log n). Jadi Merge Sort bukan algoritma in-place.

**Mengapa Binary Search butuh data terurut?** Binary Search memutuskan membuang separuh kiri atau kanan hanya dengan membandingkan elemen tengah. Keputusan ini hanya benar jika semua elemen di kiri pasti lebih kecil dan di kanan pasti lebih besar. Pada data acak, target bisa ada di bagian yang dibuang sehingga hasilnya salah.

**Kapan biaya sorting layak dibayar?** Sorting sekali membutuhkan O(n log n). Jika pencarian dilakukan berulang kali (k kali), total Linear Search = O(k*n), sedangkan sort + Binary Search = O(n log n + k log n). Sorting layak jika jumlah pencarian cukup banyak (k besar) dan data jarang berubah. Jika hanya mencari satu kali atau data terus berubah, Linear Search lebih hemat.

# SOAL 2 - Sistem Antrean Layanan dengan Undo (Linked Queue + Stack)

## Nomor 1 (Tugas 1) - Implementasi class Node, LinkedQueue, Stack

```python
# ============================================================
# CLASS NODE
# ============================================================
class Node:
    """Satu simpul pada singly linked list."""
    def __init__(self, data):
        self.data = data                  # isi simpul
        self.next = None                  # penunjuk ke simpul berikutnya


# ============================================================
# CLASS LINKEDQUEUE (singly linked list: front, rear, count)
# ============================================================
class LinkedQueue:
    def __init__(self):
        self.front = None                 # simpul paling depan (tempat dequeue)
        self.rear = None                  # simpul paling belakang (tempat enqueue)
        self.count = 0                    # jumlah elemen

    def is_empty(self):
        return self.front is None         # kosong jika front tidak menunjuk apa pun

    def size(self):
        return self.count

    def enqueue(self, data):
        """Tambah di BELAKANG (rear). O(1) karena ada referensi rear."""
        baru = Node(data)
        if self.is_empty():               # kasus antrean kosong
            self.front = baru
            self.rear = baru
        else:
            self.rear.next = baru         # sambungkan rear lama ke node baru
            self.rear = baru              # pindahkan rear
        self.count += 1

    def enqueue_front(self, data):
        """Tambah di DEPAN (front). Dipakai oleh UNDO. O(1)."""
        baru = Node(data)
        baru.next = self.front            # node baru menunjuk ke front lama
        self.front = baru                 # front pindah ke node baru
        if self.rear is None:             # jika sebelumnya kosong, rear juga = node baru
            self.rear = baru
        self.count += 1

    def dequeue(self):
        """Ambil dari DEPAN. O(1). Aman jika kosong (tidak error)."""
        if self.is_empty():
            print("  [PESAN] Dequeue gagal: Queue kosong.")
            return None
        data = self.front.data            # simpan data front
        self.front = self.front.next      # geser front ke node berikutnya
        if self.front is None:            # jika jadi kosong, rear harus di-reset juga
            self.rear = None
        self.count -= 1
        return data

    def peek(self):
        """Lihat elemen paling depan tanpa menghapus. O(1)."""
        if self.is_empty():
            print("  [PESAN] Peek gagal: Queue kosong.")
            return None
        return self.front.data

    def display(self):
        """Tampilkan isi dari front ke rear."""
        if self.is_empty():
            return "(kosong)"
        isi = []
        cur = self.front
        while cur is not None:            # telusuri dari front sampai akhir
            isi.append(str(cur.data))
            cur = cur.next
        return " -> ".join(isi)


# ============================================================
# CLASS STACK (Python list sebagai penyimpanan internal)
# ============================================================
class Stack:
    def __init__(self):
        self._items = []                  # penyimpanan internal (akhir list = top)

    def push(self, data):
        self._items.append(data)          # O(1) amortized

    def pop(self):
        """Ambil elemen teratas. Aman jika kosong."""
        if self.is_empty():
            print("  [PESAN] Pop gagal: Stack kosong.")
            return None
        return self._items.pop()          # pop() dari akhir list = O(1)

    def peek(self):
        if self.is_empty():
            print("  [PESAN] Peek gagal: Stack kosong.")
            return None
        return self._items[-1]

    def is_empty(self):
        return len(self._items) == 0

    def size(self):
        return len(self._items)

    def display(self):
        """Tampilkan dari bawah ke atas (top ada di paling kanan)."""
        if self.is_empty():
            return "(kosong)"
        return " | ".join(str(x) for x in self._items) + "   <- top"

print("Class Node, LinkedQueue, dan Stack siap dipakai.")
```

```python
# ============================================================
# UJI OPERASI DASAR sebelum simulasi utama
# ============================================================
print("--- Uji dasar LinkedQueue ---")
q_uji = LinkedQueue()
print("is_empty awal :", q_uji.is_empty())
q_uji.enqueue("X1"); q_uji.enqueue("X2"); q_uji.enqueue("X3")
print("Isi setelah enqueue X1,X2,X3 :", q_uji.display(), "| size =", q_uji.size())
q_uji.enqueue_front("X0")
print("Isi setelah enqueue_front X0 :", q_uji.display())
print("peek  :", q_uji.peek())
print("dequeue:", q_uji.dequeue(), "->", q_uji.display(), "| size =", q_uji.size())

print("\n--- Uji dasar Stack ---")
s_uji = Stack()
s_uji.push("S1"); s_uji.push("S2"); s_uji.push("S3")
print("Isi setelah push S1,S2,S3 :", s_uji.display(), "| size =", s_uji.size())
print("peek :", s_uji.peek())
print("pop  :", s_uji.pop(), "->", s_uji.display())
print("is_empty:", s_uji.is_empty())
```

## Nomor 2 (Tugas 2) - Simulasi 12 operasi

```python
# ============================================================
# SIMULASI 12 OPERASI
# ============================================================
# ID layanan dibentuk dari d (digit terakhir NIM)
d = int(NIM[-1])
A = f'MHS-{d}01'
B = f'MHS-{d}02'
C = f'MHS-{d}03'
D = f'MHS-{d}04'
E = f'MHS-{d}05'
F = f'MHS-{d}06'

queue = LinkedQueue()     # antrean aktif
riwayat = Stack()         # riwayat mahasiswa yang sudah dilayani

def tampilkan_status():
    """Cetak isi Queue (front->rear) dan Stack setelah tiap operasi."""
    print("  Queue :", queue.display())
    print("  Stack :", riwayat.display())

def jalankan(langkah, op, x=None):
    """Menjalankan satu operasi sesuai aturan soal."""
    judul = f"{op} {x}" if x else op
    print(f"Langkah {langkah}: {judul}")

    if op == "DATANG":
        queue.enqueue(x)                  # DATANG X -> enqueue(X)

    elif op == "LAYANI":
        dilayani = queue.dequeue()        # ambil dari front
        if dilayani is not None:          # jika berhasil -> catat di riwayat
            riwayat.push(dilayani)
            print(f"  (dilayani: {dilayani})")

    elif op == "UNDO":
        kembali = riwayat.pop()           # ambil riwayat terakhir (LIFO)
        if kembali is not None:           # jika ada -> kembalikan ke DEPAN antrean
            queue.enqueue_front(kembali)
            print(f"  (undo: {kembali} dikembalikan ke depan antrean)")
        else:
            print("  [PESAN] UNDO dibatalkan: tidak ada riwayat.")

    tampilkan_status()
    print()

# Urutan operasi persis seperti tabel soal
operasi = [
    ("DATANG", A),   # 1
    ("DATANG", B),   # 2
    ("DATANG", C),   # 3
    ("LAYANI", None),# 4
    ("DATANG", D),   # 5
    ("LAYANI", None),# 6
    ("DATANG", E),   # 7
    ("UNDO", None),  # 8
    ("LAYANI", None),# 9
    ("LAYANI", None),# 10
    ("UNDO", None),  # 11
    ("DATANG", F),   # 12
]

for nomor, (op, x) in enumerate(operasi, start=1):
    jalankan(nomor, op, x)
```

## Nomor 5 (Tugas 5) - FIFO vs LIFO (contoh dari simulasi)

Misal d = 7 (A = MHS-701, B = MHS-702, ...).

**Queue = FIFO (First In First Out).** Pada langkah 1-3 urutan datang adalah A, B, C. Saat LAYANI pada langkah 4, yang keluar adalah **A (MHS-701)**, yaitu yang datang paling awal, bukan C. Pada langkah 6, B yang dilayani berikutnya. Jadi urutan keluar sama dengan urutan masuk.

**Stack = LIFO (Last In First Out).** Setelah A dan B dilayani, riwayat berisi A lalu B (top = B). Saat UNDO pada langkah 8, yang di-pop adalah **B**, yaitu yang terakhir masuk ke stack, bukan A. Selanjutnya B ditaruh kembali di depan Queue lewat `enqueue_front`, sehingga langkah 9 melayani B lagi. Ini menunjukkan elemen terakhir yang masuk adalah yang pertama keluar.

## Nomor 6 (Tugas 6) - Kompleksitas waktu

| Struktur | Operasi | Kompleksitas | Alasan |
|---|---|---|---|
| LinkedQueue | enqueue | O(1) | langsung sambung ke `rear`, tanpa menelusuri list |
| LinkedQueue | enqueue_front | O(1) | node baru dihubungkan ke `front` |
| LinkedQueue | dequeue | O(1) | cukup geser `front` ke `front.next` |
| LinkedQueue | peek | O(1) | langsung baca `front.data` |
| Stack | push | O(1) (amortized) | `list.append` di akhir |
| Stack | pop | O(1) | `list.pop()` dari akhir |
| Stack | peek | O(1) | akses `_items[-1]` |

**Peran referensi front dan rear:** `front` memberi akses langsung ke titik keluar (dequeue/peek) dan `rear` memberi akses langsung ke titik masuk (enqueue). Tanpa `rear`, enqueue harus menelusuri seluruh node dari front sampai akhir (O(n)). Dengan kedua referensi, semua operasi utama Queue cukup mengubah beberapa pointer sehingga O(1), berapa pun panjang antrean.

## Nomor 7 (Tugas 7) - Mengapa linked list lebih tepat daripada `list.pop(0)`

`list.pop(0)` menghapus elemen pertama lalu **menggeser semua elemen sisanya satu posisi ke kiri**, sehingga biayanya O(n). Pada antrean yang sering melakukan dequeue, biaya ini terakumulasi: n kali dequeue menjadi O(n^2) total.

Pada linked queue, dequeue hanya memindahkan `front = front.next` (O(1)) tanpa menggeser data, sehingga n kali dequeue hanya O(n) total. Linked list juga memungkinkan `enqueue_front` untuk fitur Undo dalam O(1), sedangkan `list.insert(0, x)` pada Python list juga O(n). Karena itu linked list lebih efisien dan lebih sesuai untuk antrean dengan dequeue yang sering.
